Q10. Split the given NumPy array into three parts using `np.split()`

```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 99, 99, 3, 2, 1])
print("\nOriginal Array: ", arr)
split = np.split(arr, [3,6]) #Splits array at index 3 and 6
print("\nParted Array: ", split)
for part in split:
    print(part)
```
