#### Q5. Perform array slicing in NumPy with an example using a 1D array

Array slicing in NumPy is used to access a specific portion of an array by specifying a range of indices.

Syntax: [start:stop:step]

```python
import numpy as np
import random
```

```python
arr = np.array([10, 20, 30, 40, 50, 60, 70, 80])
print("Original Array: ", arr)
print("First 4 elements: ", arr[:4])
print("Elements from index 2 to 5: ", arr[2:6])
print("Every 2nd element: ", arr[::2])
print("Reverse: ", arr[::-1])
```
