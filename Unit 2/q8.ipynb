Q8. Create a matrix with shape X(2,3,4) and give the output of the following expressions.
Given expressions:
1. `x[0]`
2. `x[:, 1, :]`
3. `x[1, :, 2]`
4. `x[::-1, :, :]`
5. `x[:, :, -1]`

```python
import numpy as np
```

Using `arange` Attribute syntax: `np.arange(start:stop:step)` this creates a array of given length
example:  `np.arange(1,5)` creates a array of values 1 to 4

```python
x = np.arange(1, 25).reshape(2,3,4) # 2 x 3 x 4 = 24
print("Original Array:\n", x)
```

```python
print("\nx[0]: \n",x[0])
print("\nx[:, 1, :]:\n",x[:, 1, :])
print("\nx[1,:,2]: ", x[1,:,2])
print("\nx[::-1, :, :]:\n", x[::-1, :, :])
print("\nx[:,:,-1]:\n", x[:,:,-1])
```
