#### Q4. Explain the attributes of a NumPy array like ndim, shape, size, dtype, and nbytes with a code example.

##### NumPy provides several attributes to obtain information about an array, such as its dimensions, shape, number of elements, data type and memory consumption.

```python
import numpy as np
import random
```

```python
arr = np.random.random((3,2))
print(arr)
```

```python
print("Dimension: ", arr.ndim)
print("\nShape: ", arr.shape)
print("\nSize: ", arr.size)
print("\nDatatype: ", arr.dtype)
```
