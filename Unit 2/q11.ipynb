Q11. Replace missing values (NaN) with the mean and normalize the array.

Question: Write a NumPy program to replace missing values with the mean and normalize the data between 0 and 1.

```python
import numpy as np
```

```python
import numpy as np

arr = np.array([10, 20, np.nan, 40, 50])

# Calculate mean, ignoring NaN
mean = np.nanmean(arr) #Mean = (10 + 20 + 40 + 50) / 4 = 30

# Replace NaN with mean = 30
arr[np.isnan(arr)] = mean

# Normalize data between 0 and 1
normalized = (arr - np.min(arr)) / (np.max(arr) - np.min(arr)) #Scales values between 0 and 1
```

Formula: $$X_{normalized} = \frac{X-X_{min}}{X_{max}-X_{min}}$$

```python
print("Mean:", mean)
print("After replacing NaN:", arr)
print("Normalized array:", normalized)
```
