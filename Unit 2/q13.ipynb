Q13. Perform data manipulation on student scores using NumPy.

Question: Write a full program that creates an array of student scores and performs the following operations:

1. Replace missing scores with the average.
2. Filter scores above a threshold.
3. Reshape into 2 groups.
4. Normalize scores.

```python
import numpy as np
```

```python
scores = np.array([
    45, 60, np.nan, 80,
    90, 55, 70, 85
]) # with a missing value (np.nan)
```

1. Replacing Missing values with average

```python
mean = np.nanmean(scores)
scores[np.isnan(scores)] = mean
print("Scores after replacing missing values:")
print(scores)
```

2. Filter Score above 60

```python
filtered = scores[scores > 60]
print("\n Scores above 60: ")
print(filtered)
```

3. Reshape into 2 groups

```python
groups = scores.reshape(2,4)
print("\nScores divided into 2 groups:")
print(groups)
```

4. Normalize scores between 0 and 1

Formula: $$X_{normalized} = \frac{X-X_{min}}{X_{max}-X_{min}}$$

Here $X = scroes$


```python
normalized = (scores - np.min(scores))/(np.max(scores) - np.min(scores))
print("\nNormalized scores:")
print(normalized)
```
