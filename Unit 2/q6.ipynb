#### Q6. What does the .reshape() function do in NumPy? Illustrate with an example.

The reshape() function in NumPy is used to change the shape of an array without changing its original data.

It can convert a 1D array into a 2D or 3D array, provided the total number of elements remains the same.
Syntax:
`
array.reshape(rows, columns)
`

```python
import numpy as np
```

```python
arr = np.array([10,20,30,40,50,60])
```

```python
print("Original Array: ",arr)
print("Dimension of array: ", arr.ndim)
```

```python
reshaped_array = arr.reshape(3,2)
print("Reshaped Array 3 X 2 Matrix \n", reshaped_array)
```
