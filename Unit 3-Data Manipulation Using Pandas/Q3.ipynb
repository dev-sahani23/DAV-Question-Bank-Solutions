Q3. Create a Pandas Series from a list and display its index and values.

```python
import pandas as pd
```

```python
# Create a list
data = [10, 20, 30, 40, 50]

# Create a Pandas Series
s = pd.Series(data)

# Display Series
print("Pandas Series:")
print(s)

# Display index and values
print("\nIndex:")
print(s.index)

print("\nValues:")
print(s.values)
```
